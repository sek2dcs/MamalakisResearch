```python
import os
import numpy as np
import netCDF4 as nc
import warnings

def get_model_delta(model_list, scenario, var_idx, target_start, target_end, 
                    base_start=2015, base_end=2024, single_model_idx=None):
    """
    var_idx: Integer (0-6) representing which of the 7 variables to analyze
    single_model_idx: If an integer is passed, only that model in model_list is processed.
    """
    
    # 1. Determine which models to process
    if single_model_idx is not None:
        processing_list = [model_list[single_model_idx]]
    else:
        processing_list = model_list

    results = []

    for filename in processing_list:
        full_path = os.path.join(data_path, filename)
        with nc.Dataset(full_path) as ds:
            # Shape: (ensemble, variable, time, lat, lon)
            # We select ONLY the variable we want (var_idx) to save memory
            data = ds[f"data_{scenario}"][:, var_idx, :, :, :]
            
            # --- BASE PERIOD (2015-2024) ---
            b_start = (base_start - 2015) * 12
            b_end = (base_end - 2015 + 1) * 12
            
            # Average over Ensembles (axis 0) AND Time (axis 1)
            with warnings.catch_warnings():
                warnings.simplefilter("ignore", category=RuntimeWarning)
                base_slice = data[:, b_start:b_end, :, :]
                base_mean = np.nanmean(base_slice, axis=(0, 1)) 
            
            # --- TARGET PERIOD ---
            t_start = (target_start - 2015) * 12
            t_end = (target_end - 2015 + 1) * 12
            
            with warnings.catch_warnings():
                warnings.simplefilter("ignore", category=RuntimeWarning)
                target_slice = data[:, t_start:t_end, :, :]
                target_mean = np.nanmean(target_slice, axis=(0, 1))

            # --- CALCULATE DELTA (Anomaly) ---
            delta = target_mean - base_mean
            results.append(delta)

    # If processing multiple models, you can return the average of all models
    final_output = np.nanmean(results, axis=0)
    
    # Land mask cleanup
    final_output = np.nan_to_num(final_output, nan=0.0)
    
    return final_output
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_delta_map(delta_map, title="Climate Anomaly"):
    # If shape is (144, 73), transpose to (73, 144) 
    # so Latitude is on the Y-axis and Longitude is on the X-axis
    if delta_map.shape == (144, 73):
        plot_data = delta_map.T
    else:
        plot_data = delta_map

    plt.figure(figsize=(12, 6))
    
    # Create coordinate arrays based on the 2.5 degree resolution
    lons = np.linspace(0, 360, plot_data.shape[1])
    lats = np.linspace(-90, 90, plot_data.shape[0])
    
    # Use 'RdBu_r' (Red-Blue reversed) colormap 
    # Red = warming/increase, Blue = cooling/decrease
    # centered at 0 using vmin/vmax
    limit = np.nanmax(np.abs(plot_data))
    
    mesh = plt.pcolormesh(lons, lats, plot_data, 
                          cmap='RdBu_r', 
                          shading='auto',
                          vmin=-limit, vmax=limit)
    
    plt.colorbar(mesh, label='Anomaly Value')
    plt.title(title)
    plt.xlabel('Longitude')
    plt.ylabel('Latitude')
    plt.grid(linestyle='--', alpha=0.5)
    
    plt.show()

# Example usage:
plot_delta_map(delta_map, title="Temperature Change (2050-2059 vs 2015-2024)")
```

```python
import os
import numpy as np
import netCDF4 as nc
import warnings

def get_model_tensors(model_list, scenario, data_path, var_idx, target_start, target_end, 
                      base_start=2015, base_end=2024, single_model_idx=None, subtract_base=False):
    """
    Returns x_data (samples) and y_data (labels) for CNN training.
    subtract_base: If True, returns (Target - Base). If False, returns both periods as separate samples.
    """
    
    if single_model_idx is not None:
        processing_list = [model_list[single_model_idx]]
    else:
        processing_list = model_list

    x_list = []
    y_list = []

    for filename in processing_list:
        full_path = os.path.join(data_path, filename)
        with nc.Dataset(full_path) as ds:
            # Shape: (ensemble, variable, time, lat, lon)
            data = ds[f"data_{scenario}"][:, var_idx, :, :, :]
            
            # --- BASE PERIOD (Label 0) ---
            b_start = (base_start - 2015) * 12
            b_end = (base_end - 2015 + 1) * 12
            
            # --- TARGET PERIOD (Label 1) ---
            t_start = (target_start - 2015) * 12
            t_end = (target_end - 2015 + 1) * 12

            for ens_idx in range(5):
                with warnings.catch_warnings():
                    warnings.simplefilter("ignore", category=RuntimeWarning)
                    
                    # Annual Mean for this ensemble in Base Period
                    base_ann = np.nanmean(data[ens_idx, b_start:b_end, :, :], axis=0)
                    # Annual Mean for this ensemble in Target Period
                    target_ann = np.nanmean(data[ens_idx, t_start:t_end, :, :], axis=0)

                if subtract_base:
                    # Case 1: You want the difference (Anomaly) as the input
                    x_list.append(target_ann - base_ann)
                    y_list.append(1) # Usually, deltas are treated as 'future' signal
                else:
                    # Case 2: You want both periods as individual training samples
                    x_list.append(base_ann)
                    y_list.append(0)
                    
                    x_list.append(target_ann)
                    y_list.append(1)

    # Convert to numpy arrays
    x_data = np.array(x_list)
    y_data = np.array(y_list).reshape(-1, 1)

    # Add the "Channel" dimension for the CNN: (Samples, Lat, Lon) -> (Samples, 1, Lat, Lon)
    x_data = np.expand_dims(x_data, axis=1)
    
    # Cleanup NaNs (Land Masks)
    x_data = np.nan_to_num(x_data, nan=0.0)
    
    return x_data, y_data
```

```python
import os
import numpy as np
import netCDF4 as nc
import warnings

def get_cnn_tensors(model_list, scenario, data_path, var_idx, 
                    st_early=2015, end_early=2024, 
                    st_late=2050, end_late=2059, 
                    models_to_run=None):
    """
    models_to_run: 
        - None: Runs all models in model_list
        - int: Runs only the model at that index (e.g., 0)
        - list of ints: Runs a subset (e.g., [0, 2, 4])
    """
    
    # 1. Determine which models to process based on input
    if models_to_run is None:
        selected_models = model_list
    elif isinstance(models_to_run, int):
        selected_models = [model_list[models_to_run]]
    elif isinstance(models_to_run, list):
        selected_models = [model_list[i] for i in models_to_run]
    else:
        selected_models = model_list

    x_list = []
    y_list = []

    # Define the two comparison periods
    periods = [(st_early, end_early, 0), (st_late, end_late, 1)]

    print(f"Processing {len(selected_models)} model(s)...")

    for filename in selected_models:
        full_path = os.path.join(data_path, filename)
        
        with nc.Dataset(full_path) as ds:
            # Load specific variable: [Ensemble, Var, Time, Lat, Lon]
            # Slicing var_idx collapses that dimension
            data = ds[f"data_{scenario}"][:, var_idx, :, :, :]
            
            for start_yr, end_yr, label in periods:
                for yr_idx in range(start_yr, end_yr + 1):
                    # Calculate 0-based index (assuming files start at Jan 2015)
                    month_start = (yr_idx - 2015) * 12
                    month_end = month_start + 12
                    
                    for ens_idx in range(5):
                        annual_slice = data[ens_idx, month_start:month_end, :, :]
                        
                        with warnings.catch_warnings():
                            warnings.simplefilter("ignore", category=RuntimeWarning)
                            # Average the 12 months -> (144, 73)
                            annual_mean = np.nanmean(annual_slice, axis=0)
                        
                        x_list.append(annual_mean)
                        y_list.append(label)

    # Convert to arrays and reshape
    x_data = np.array(x_list)
    y_data = np.array(y_list).reshape(-1, 1)

    # Add Channel dimension for CNN: (Samples, 1, 144, 73)
    x_data = np.expand_dims(x_data, axis=1)
    
    # Land Mask cleanup: convert NaNs to 0.0
    x_data = np.nan_to_num(x_data, nan=0.0)
    
    print(f"Final Shape: {x_data.shape}, Labels: {y_data.shape}")
    return x_data, y_data
```

```python
import os
import numpy as np
import netCDF4 as nc
import warnings

def get_cnn_tensors(model_list, scenario, data_path, var_idx, 
                    st_early=2015, end_early=2024, 
                    st_late=2050, end_late=2059,
                    stat='mean', use_anomaly=True, 
                    base_start=2015, base_end=2024,
                    models_to_run=None):
    """
    stat: 'mean', 'std', 'max', 'min', or 'median'
    use_anomaly: If True, subtracts the base period's stat from the yearly stat.
    """
    
    # Map string inputs to numpy functions
    stat_map = {
        'mean': np.nanmean,
        'std': np.nanstd,
        'max': np.nanmax,
        'min': np.nanmin,
        'median': np.nanmedian
    }
    
    # Default to mean if input is unrecognized
    calc_func = stat_map.get(stat.lower(), np.nanmean)

    if models_to_run is None:
        selected_models = model_list
    elif isinstance(models_to_run, int):
        selected_models = [model_list[models_to_run]]
    else:
        selected_models = [model_list[i] for i in models_to_run]

    x_list = []
    y_list = []
    periods = [(st_early, end_early, 0), (st_late, end_late, 1)]

    print(f"Generating tensors | Stat: {stat} | Anomaly: {use_anomaly}")

    for filename in selected_models:
        full_path = os.path.join(data_path, filename)
        with nc.Dataset(full_path) as ds:
            data = ds[f"data_{scenario}"][:, var_idx, :, :, :]
            
            # 1. Calculate base period baseline for this specific stat
            if use_anomaly:
                b_start = (base_start - 2015) * 12
                b_end = (base_end - 2015 + 1) * 12
                with warnings.catch_warnings():
                    warnings.simplefilter("ignore", category=RuntimeWarning)
                    # Result shape: (5, 144, 73)
                    ensemble_base_baseline = calc_func(data[:, b_start:b_end, :, :], axis=1)

            # 2. Process yearly samples
            for start_yr, end_yr, label in periods:
                for yr_idx in range(start_yr, end_yr + 1):
                    m_start = (yr_idx - 2015) * 12
                    m_end = m_start + 12
                    
                    for ens_idx in range(5):
                        annual_slice = data[ens_idx, m_start:m_end, :, :]
                        
                        with warnings.catch_warnings():
                            warnings.simplefilter("ignore", category=RuntimeWarning)
                            # Apply the chosen statistic (mean, std, etc.)
                            yearly_val = calc_func(annual_slice, axis=0)
                        
                        if use_anomaly:
                            output_map = yearly_val - ensemble_base_baseline[ens_idx]
                        else:
                            output_map = yearly_val
                        
                        x_list.append(output_map)
                        y_list.append(label)

    # Convert to standard CNN format
    x_data = np.array(x_list)
    y_data = np.array(y_list).reshape(-1, 1)
    x_data = np.expand_dims(x_data, axis=1) # (Samples, 1, 144, 73)
    x_data = np.nan_to_num(x_data, nan=0.0)
    
    return x_data, y_data
```

```python
import numpy as np

def convert_units(varname: str, x: np.ndarray):
    """
    Convert raw units into nicer units for CNN training and plotting.
    """
    if varname in {"tas", "tasmax", "tasmin"}:
        # Kelvin to Celsius
        return x - 273.15, "$^{\circ}$C"
    if varname == "pr":
        # kg/m2/s to mm/day
        return x * 86400.0, "mm/day"
    if varname == "psl":
        # Pascals to hPa
        return x / 100.0, "hPa"
    if varname == "sfcWind":
        return x, "m/s"
    if varname == "mrsos":
        return x, "kg/m$^{2}$"
    return x, "unknown"
```

```python
import os
import netCDF4 as nc
import warnings

def get_cnn_tensors(model_list, scenario, data_path, var_idx, 
                    st_early=2015, end_early=2024, 
                    st_late=2050, end_late=2059,
                    stat='mean', use_anomaly=True, 
                    base_start=2015, base_end=2024,
                    models_to_run=None):
    
    # 1. Map your indices to your specific variable list
    var_list = ["tas", "tasmax", "tasmin", "pr", "psl", "sfcWind", "mrsos"]
    varname = var_list[var_idx] if var_idx < len(var_list) else "unknown"

    # 2. Select statistic function
    stat_map = {
        'mean': np.nanmean, 'std': np.nanstd, 
        'max': np.nanmax, 'min': np.nanmin, 'median': np.nanmedian
    }
    calc_func = stat_map.get(stat.lower(), np.nanmean)

    # 3. Model subsetting
    if models_to_run is None:
        selected_models = model_list
    elif isinstance(models_to_run, int):
        selected_models = [model_list[models_to_run]]
    else:
        selected_models = [model_list[i] for i in models_to_run]

    x_list, y_list = [], []
    periods = [(st_early, end_early, 0), (st_late, end_late, 1)]

    print(f"Variable: {varname} | Stat: {stat} | Anomaly: {use_anomaly}")

    for filename in selected_models:
        full_path = os.path.join(data_path, filename)
        with nc.Dataset(full_path) as ds:
            # Extract raw data [Ensemble, Time, Lat, Lon]
            data_raw = ds[f"data_{scenario}"][:, var_idx, :, :, :]
            
            # --- UNIT CONVERSION ---
            data, unit_str = convert_units(varname, data_raw)
            
            if use_anomaly:
                b_start, b_end = (base_start-2015)*12, (base_end-2015+1)*12
                with warnings.catch_warnings():
                    warnings.simplefilter("ignore", category=RuntimeWarning)
                    # Base for anomaly is calculated from converted data
                    ensemble_base = calc_func(data[:, b_start:b_end, :, :], axis=1)

            for start_yr, end_yr, label in periods:
                for yr_idx in range(start_yr, end_yr + 1):
                    m_start, m_end = (yr_idx-2015)*12, (yr_idx-2015)*12 + 12
                    
                    for ens_idx in range(5):
                        annual_slice = data[ens_idx, m_start:m_end, :, :]
                        
                        with warnings.catch_warnings():
                            warnings.simplefilter("ignore", category=RuntimeWarning)
                            yearly_val = calc_func(annual_slice, axis=0)
                        
                        # Apply Anomaly logic
                        val = yearly_val - ensemble_base[ens_idx] if use_anomaly else yearly_val
                        x_list.append(val)
                        y_list.append(label)

    # Final array formatting
    x_data = np.expand_dims(np.array(x_list), axis=1) # (Samples, 1, 144, 73)
    y_data = np.array(y_list).reshape(-1, 1)
    
    return np.nan_to_num(x_data, nan=0.0), y_data
```

```python
def plot_x_sample(x_data, y_data, sample_idx=0, title=None, units=""):
    # Extract sample and remove channel dim (1, 144, 73) -> (144, 73)
    # Transpose to (73, 144) for correct Lat/Lon orientation
    plot_data = x_data[sample_idx, 0].T
    
    plt.figure(figsize=(12, 5))
    
    # Use a diverging map for anomalies (centered at 0)
    # or a sequential map for absolute values
    is_anomaly = np.any(plot_data < 0) and np.any(plot_data > 0)
    cmap = 'RdBu_r' if is_anomaly else 'viridis'
    
    # Set color limits to be symmetric if it's an anomaly
    if is_anomaly:
        limit = np.nanmax(np.abs(plot_data))
        vmin, vmax = -limit, limit
    else:
        vmin, vmax = None, None

    img = plt.imshow(plot_data, origin='lower', cmap=cmap, 
                     vmin=vmin, vmax=vmax, aspect='auto',
                     extent=[0, 360, -90, 90])
    
    plt.colorbar(img, label=f'Value ({units})')
    
    label_text = "Early Period" if y_data[sample_idx] == 0 else "Late Period"
    plt.title(title if title else f"Sample {sample_idx} | {label_text}")
    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.show()
```

```python
def plot_period_comparison(x_data, y_data, units=""):
    # 1. Identify indices for each label
    early_indices = np.where(y_data == 0)[0]
    late_indices = np.where(y_data == 1)[0]
    
    # 2. Calculate the mean map for each group
    # Shape of x_data is (Samples, 1, 144, 73)
    # We average over axis 0 (Samples)
    early_mean = np.mean(x_data[early_indices], axis=0)[0]
    late_mean = np.mean(x_data[late_indices], axis=0)[0]
    
    # 3. Calculate the delta (The Signal)
    signal = late_mean - early_mean
    
    # 4. Plotting
    fig, axes = plt.subplots(1, 3, figsize=(20, 5))
    titles = ["Mean Early (2015-2024)", "Mean Late (2050-2059)", "The Signal (Late - Early)"]
    maps = [early_mean.T, late_mean.T, signal.T]
    
    for i in range(3):
        # Use RdBu_r for the signal to highlight changes
        cmap = 'RdBu_r' if i == 2 else 'viridis'
        
        # Center the colorbar for the signal map
        if i == 2:
            limit = np.nanmax(np.abs(maps[i]))
            im = axes[i].imshow(maps[i], origin='lower', cmap=cmap, vmin=-limit, vmax=limit, aspect='auto')
        else:
            im = axes[i].imshow(maps[i], origin='lower', cmap=cmap, aspect='auto')
            
        axes[i].set_title(titles[i])
        plt.colorbar(im, ax=axes[i], label=f"Value ({units})")
        axes[i].set_xlabel("Longitude")
        axes[i].set_ylabel("Latitude")

    plt.tight_layout()
    plt.show()

# Usage:
plot_period_comparison(x_data, y_data, units="°C")
```

```python
# The standard order of your variables
VAR_LIST = ["tas", "tasmax", "tasmin", "pr", "psl", "sfcWind", "mrsos"]

# The corresponding units after conversion
UNIT_MAP = {
    "tas": "°C", 
    "tasmax": "°C", 
    "tasmin": "°C", 
    "pr": "mm/day", 
    "psl": "hPa", 
    "sfcWind": "m/s", 
    "mrsos": "kg/m²"
}
```

```python
def plot_period_comparison(x_data, y_data, var_idx):
    # Get the name and unit automatically
    var_name = VAR_LIST[var_idx]
    units = UNIT_MAP.get(var_name, "units")
    
    early_indices = np.where(y_data == 0)[0]
    late_indices = np.where(y_data == 1)[0]
    
    early_mean = np.mean(x_data[early_indices], axis=0)[0]
    late_mean = np.mean(x_data[late_indices], axis=0)[0]
    signal = late_mean - early_mean
    
    # Calculate symmetric limits for consistent colorbars
    all_vals = [early_mean, late_mean, signal]
    abs_max = max(abs(np.nanmin(all_vals)), abs(np.nanmax(all_vals)))

    fig, axes = plt.subplots(1, 3, figsize=(22, 6))
    titles = [f"Mean Early ({var_name})", f"Mean Late ({var_name})", "The Signal (Late - Early)"]
    maps = [early_mean.T, late_mean.T, signal.T]
    
    for i in range(3):
        im = axes[i].imshow(maps[i], origin='lower', cmap='RdBu_r', 
                            vmin=-abs_max, vmax=abs_max, aspect='auto')
        
        axes[i].set_title(titles[i])
        plt.colorbar(im, ax=axes[i], label=f"{var_name} ({units})")
        axes[i].set_xlabel("Longitude Index")
        axes[i].set_ylabel("Latitude Index")

    plt.tight_layout()
    plt.show()
```
